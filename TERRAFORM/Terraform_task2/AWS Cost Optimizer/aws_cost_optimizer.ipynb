# AWS Cost Optimization Dashboard

```python
import boto3
from botocore.exceptions import NoCredentialsError, ClientError

try:
    ec2_client = boto3.client('ec2', region_name='ap-south-1')
    print("[INFO] AWS EC2 client initialized successfully for ap-south-1.")
except Exception as e:
    print(f"[ERROR] Client initialization failed: {str(e)}")
```

```python
try:
    print("[INFO] Querying AWS infrastructure for Elastic IP addresses...")
    addresses_dict = ec2_client.describe_addresses()
    ip_list = addresses_dict.get('Addresses', [])
    print(f"[SUCCESS] Retrieved {len(ip_list)} Elastic IP resource(s) from the account.")
except NoCredentialsError:
    print("[ERROR] AWS credentials not found. Please run 'aws configure' in your environment.")
except ClientError as e:
    print(f"[ERROR] AWS API Client Error: {e.response['Error']['Message']}")
except NameError:
    print("[ERROR] EC2 client not initialized. Please run Cell 2 first.")
```

```python
try:
    print("[INFO] Querying AWS infrastructure for Elastic IP addresses...")
    addresses_dict = ec2_client.describe_addresses()
    ip_list = addresses_dict.get('Addresses', [])
    print(f"[SUCCESS] Retrieved {len(ip_list)} Elastic IP resource(s) from the account.")
except NoCredentialsError:
    print("[ERROR] AWS credentials not found. Please run 'aws configure' in your environment.")
except ClientError as e:
    print(f"[ERROR] AWS API Client Error: {e.response['Error']['Message']}")
except NameError:
    print("[ERROR] EC2 client not initialized. Please run Cell 2 first.")
```

```python
unattached_ips = []

try:
    for eip in ip_list:
        if 'AssociationId' not in eip:
            unattached_ips.append({
            'PublicIp': eip['PublicIp'],
        'AllocationId': eip['AllocationId']
            })
        print(f"[WARNING] Idle resource detected: {eip['PublicIp']}")
            
    if not unattached_ips and 'ip_list' in locals():
        print("[INFO] Processing complete. No unattached Elastic IPs found in the current pool.")
except NameError:
    print("[ERROR] IP list data missing. Please execute Cell 3 first.")
```

```python
try:
    print("=" * 60)
    print("VPC INFRASTRUCTURE COST OPTIMIZATION REPORT")
    print("=" * 60)
    print(f"Total Elastic IPs Evaluated : {len(ip_list)}")
    print(f"Total Idle Elastic IPs Found: {len(unattached_ips)}")
    print("=" * 60)

    if len(unattached_ips) > 0:
        print(f"[ACTION REQUIRED] Release the {len(unattached_ips)} idle Elastic IP(s) via the AWS Console to stop accumulation charges.")
    else:
        print("[STATUS] Compliance verified. Infrastructure state is currently optimized.")
except NameError:
    print("[ERROR] Unable to generate report. Ensure Cells 3 and 4 have been executed successfully.")
```
