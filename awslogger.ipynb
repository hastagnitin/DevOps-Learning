```python
import boto3
from datetime import datetime
```

```python
AWS_ACCESS_KEY = ""
AWS_SECRET_KEY = ""
AWS_REGION = "ap-south-1"


```

```python
def generate_cloud_audit_report(access_key, secret_key, region):
    
    report_filename = "aws_infrastructure_report.txt"
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    try:
        
        s3_client = boto3.client(
            "s3",
            aws_access_key_id=access_key,
            aws_secret_access_key=secret_key,
            region_name=region,
        )

        print("[INFO] Fetching live deployment data from AWS infrastructure...")
        response = s3_client.list_buckets()
        buckets = response.get("Buckets", [])

        
        print(f"[INFO] Opening file streaming pipeline for {report_filename}...")
        with open(report_filename, "w") as report:
            report.write("-------------\n")
            report.write(f"⚙️ AUTOMATED AWS INFRASTRUCTURE REPORT\n")
            report.write(f"Generated On : {timestamp} (IST)\n")
            report.write(f"Target Region: {region}\n")
            report.write("-------------\n\n")

            report.write("[SECTION: S3 BUCKET INVENTORY]\n")
            if not buckets:
                report.write("Status: Safe. No active storage buckets detected.\n")
            else:
                report.write(f"Total Active Buckets Detected: {len(buckets)}\n")
                for index, bucket in enumerate(buckets, 1):
                    report.write(
                        f" {index}. Name: {bucket['Name']} | Created: {bucket['CreationDate']}\n"
                    )

            report.write("\n-------------\n")
            report.write("End of Automated Logging Pipeline.\n")

        print(
            f"[SUCCESS] Infrastructure logs written beautifully to '{report_filename}'!"
        )

    except Exception as error:
        print(f"\n[CRITICAL ERROR] Failed to complete cloud audit routing.")
        print(f"Diagnostics: {error}")


if __name__ == "__main__":
    generate_cloud_audit_report(AWS_ACCESS_KEY, AWS_SECRET_KEY, AWS_REGION)
```
