```python
import boto3


AWS_ACCESS_KEY = ""
AWS_SECRET_KEY = ""


def list_buckets_func():
    s3 = boto3.client(
        "s3",
        aws_access_key_id=AWS_ACCESS_KEY,
        aws_secret_access_key=AWS_SECRET_KEY,
        region_name="ap-south-1"
    )
    response = s3.list_buckets()
    
    
    bucket_names = [bucket['Name'] for bucket in response['Buckets']]
    print("\n S3 Buckets  List:", bucket_names)



def main_tool():
    
    print("""
    1. List Bucket
    2. Upload in Bucket
    3. Download From Bucket
    4. Exit
    """)
    
    
    choice = input("Enter your choice: ")
    
    
    if choice == "1":
        list_buckets_func()
    elif choice == "2":
        print("\nUpload  function ")
    elif choice == "3":
        print("\nDownload  function ")
    elif choice == "4":
        print("\n Bye!")
    else:
        print("\nInvalid choice! 1, 2, 3 ya 4 ")


main_tool()
```

```python
import boto3


AWS_ACCESS_KEY = "ENTER_YOUR_REAL_ACCESS_KEY"
AWS_SECRET_KEY = "ENTER_YOUR_REAL_SECRET_KEY"
AWS_REGION = "ap-south-1"


print("[INFO] Connecting to AWS EC2 Resource Interface...")
ec2_resource = boto3.resource(
    "ec2",
    aws_access_key_id=AWS_ACCESS_KEY,
    aws_secret_access_key=AWS_SECRET_KEY,
    region_name=AWS_REGION,
)


print("[INFO] Provisioning a new EC2 instance...")


instances = ec2_resource.create_instances(
    ImageId="ami-0db56f446d44f2f09",  
    MinCount=1,
    MaxCount=1,
    InstanceType="t2.nano",
)


new_instance_id = instances[0].id
print(f"[SUCCESS] Launched Instance ID: {new_instance_id}")



print("\n[INFO] Fetching all active instance resources in this region:")
for instance in ec2_resource.instances.all():
    print(f"🖥️ Instance Resource Object: {instance} | Active ID: {instance.id}")



print("\n[INFO] Re-initializing low-level EC2 Client for advanced lifecycle operations...")
ec2_client = boto3.client(
    "ec2",
    aws_access_key_id=AWS_ACCESS_KEY,
    aws_secret_access_key=AWS_SECRET_KEY,
    region_name=AWS_REGION,
)
```
